```python
# Função para calcular médias
def medias_escola(codigo):
    return dados_enem_limpos[
        dados_enem_limpos["CO_ESCOLA"] == codigo
    ][
        [
            'NU_NOTA_CN',
            'NU_NOTA_CH',
            'NU_NOTA_LC',
            'NU_NOTA_MT',
            'NU_NOTA_REDACAO'
        ]
    ].mean()

# Colunas de notas gerais
notas = [
    'NU_NOTA_CN',
    'NU_NOTA_CH',
    'NU_NOTA_LC',
    'NU_NOTA_MT',
    'NU_NOTA_REDACAO'
]

# Médias das escolas + geral
medias = pd.DataFrame({
    "Quarup": medias_escola(35164948.0),
    "Tecnosert": medias_escola(35125465.0),
    "Magon": medias_escola(35044635.0),

    # 🔥 ADICIONADO: MÉDIA GERAL DO ENEM
    "Geral ENEM": dados_enem_limpos[notas].mean()
}).T

# Mostrar tabela
medias
```

```python
import matplotlib.pyplot as plt

medias.plot(kind="bar", figsize=(10,6), grid=True)

plt.title("Médias das Escolas e Geral do ENEM 2024")
plt.xlabel("Escolas")
plt.ylabel("Média")

plt.tight_layout()
plt.savefig("../imagens/media_escolas.png", dpi=300)
plt.show()
```
